# Memory-Guided Question Answering

The final step combines retrieved passages and entities into an LLM prompt.

```python
%run supportvectors-common.ipynb
```

```python

from memg_concepts.config import DEFAULT_PDF
from memg_concepts.pipeline import build_index, query_index

index = build_index(DEFAULT_PDF, max_chunks=12)
questions = [
    'What problem does the Transformer solve compared to recurrent models?',
    'What are the main components of multi-head attention?',
    'How is positional information injected in the Transformer?',
]
questions
```

```python
for question in questions:
    evidence, result = query_index(index, question)
    print('=' * 80)
    print('Q:', question)
    print('Passages:', result.passage_ids)
    print('A:', result.answer)
```
