# MemGraphRAG Overview

MemGraphRAG addresses three failure modes of naive GraphRAG:

1. **Thematic irrelevance** — off-topic triples pollute the graph.
2. **Logical inconsistency** — contradictory facts appear across chunks.
3. **Structural fragmentation** — equivalent entities remain disconnected.

The framework stores knowledge in a **three-layer global memory**:

- **Schema layer** — abstract `(head_type, relation, tail_type)` patterns.
- **Fact layer** — concrete `(head_entity, relation, tail_entity)` triples.
- **Passage layer** — source text chunks that ground every fact.

We will build this pipeline on the *Attention Is All You Need* paper (`data/attention.pdf`).
